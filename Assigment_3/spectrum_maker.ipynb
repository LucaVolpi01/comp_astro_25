```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

planet_name= ["TOI-776_b" ]
#put the planet name inside list and the program save the image 
```

```python

for planet in planet_name:

    filepath = f"{planet}/outputs/{planet}_assignment3_taskB_spectrum.dat"

    df = pd.read_csv(filepath, delim_whitespace=True, header=None)

    # Extract data
    wavelength = df.iloc[:, 0]
    depth = df.iloc[:, 1]

    plt.figure(figsize=(8, 5))

    plt.errorbar(
        wavelength,
        depth,
        yerr=1e-6,
        fmt='o',
        markersize=4,
        capsize=3,
        linewidth=1,
        elinewidth=1,
        alpha=0.8
    )

    # Make error bars highly transparent
    for bar in plt.gca().collections:
        bar.set_alpha(0.25)

    plt.xlabel("Wavelength [$\mu$m]")
    plt.ylabel("Transit depth $(R_p / R_s)^2$")
    plt.title(f"Transmission Spectrum – {planet}")

    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        f"{planet}/outputs/{planet}_assignment3_taskB_spectrum.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```
